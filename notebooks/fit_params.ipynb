```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
```

```python
mpl.rcParams.update({
    "text.usetex": True,
    "font.family": "sans-serif",
    "font.serif": "cm",#["Computer Modern Roman"],
    "axes.unicode_minus": False  # ensure minus signs render correctly
})

mpl.rcParams['axes.edgecolor'] = 'black'
mpl.rcParams['axes.linewidth'] = 1.5
mpl.rcParams['xtick.direction'] = "in"
mpl.rcParams['axes.grid'] = True
mpl.rcParams['grid.alpha'] = 0.75
mpl.rcParams['grid.linestyle'] = '--'
mpl.rcParams['axes.labelsize'] = 16
mpl.rcParams['axes.titlesize'] = 16
mpl.rcParams['xtick.labelsize'] = 14
mpl.rcParams['ytick.labelsize'] = 14
mpl.rcParams['legend.fontsize'] = 12

mpl.rcParams['xtick.minor.visible'] = True
mpl.rcParams['ytick.minor.visible'] = True
mpl.rcParams['xtick.direction'] = 'in'
mpl.rcParams['ytick.direction'] = 'in'
mpl.rcParams['xtick.top'] = True
mpl.rcParams['ytick.right'] = True
mpl.rcParams['xtick.minor.top'] = True
mpl.rcParams['ytick.minor.right'] = True
mpl.rcParams['xtick.major.width'] = 1.5
mpl.rcParams['xtick.minor.width'] = 1.5
mpl.rcParams['ytick.major.width'] = 1.5
mpl.rcParams['ytick.minor.width'] = 1.5
```

```python
fit_plus = pd.read_csv(
    "../data/fit_params_b4/fitpars_plus.dat",
    skiprows=1,
    header=0,
    sep='\t'
)
fit_plus['KTcent'] = (fit_plus['KTmax'] + fit_plus['#KTmin'])/2
fit_minus = pd.read_csv(
    "../data/fit_params_b4/fitpars_minus.dat",
    skiprows=1,
    header=0,
    sep='\t'
)
fit_minus['KTcent'] = (fit_minus['KTmax'] + fit_minus['#KTmin'])/2
```

```python
fit_minus[fit_minus['KTcent']>0.2]
```

```python
fit_plus_plot = fit_plus[
    (fit_plus['fitstatus']=='converged')
    & (fit_plus['KTcent']>0.2)
    # & (fit_plus['KTcent']!=0.375)
    # & (fit_plus['KTcent']!=0.1)
]
fit_minus_plot = fit_minus[
    (fit_minus['fitstatus']=='converged')
    & (fit_minus['KTcent']>0.2)
]
plt.errorbar(
    fit_plus_plot['KTcent'], fit_plus_plot['Par0'], fit_plus_plot['Err0'],
    linestyle='', marker='o', capsize=4, color='red',
    label=r'$\pi^+\pi^+$'
)
plt.errorbar(
    fit_minus_plot['KTcent'], fit_minus_plot['Par0'], fit_minus_plot['Err0'],
    linestyle='', marker='o', capsize=4, color='red', fillstyle='none',
    label=r'$\pi^-\pi^-$'
)
plt.xlabel(r'$K_T$ [GeV]')
plt.ylabel(r'$N$')
plt.legend(frameon=False)
plt.savefig("../figures/fitParams/N.png")
```
